# Basic Numpy Functions

```python
import numpy as np
print("\n Creating a 2-D Array Using functions")
zeros_arr=np.zeros((3,4))
print("\n Array of Zeros:\n",zeros_arr)
ones_arr=np.ones((3,4))
print("\nArray of Ones:\n",ones_arr)
print("\n Identity Array:\n",np.eye(4))
reshaped_arr=np.arange(12)
print("\n Reshaped Array:\n",reshaped_arr)
Final=reshaped_arr.reshape(3,4)
print("\nFinal Array:\n",Final)
print("\nSum of elements:",np.sum(Final))
print("\nMean:",np.mean(Final))
print("\nMax Element:",np.max(Final))
```
